## **typing — Support for type hints**

### **Added in version 3.5**

Source code: `Lib/typing.py`

**Note**: The Python runtime does not enforce function and variable type annotations.  
They can be used by third-party tools such as type checkers, IDEs, linters, etc.  
This module provides runtime support for type hints.


### **Example: Function with Type Hints**

```python
def surface_area_of_cube(edge_length: float) -> str:
    return f"The surface area of the cube is {6 * edge_length ** 2:.2f}."
```

```python
result = surface_area_of_cube(4.4)
result
```

```python
type(result)
```


* The function `surface_area_of_cube` takes an argument expected to be a **float**.
* It is expected to return a **str**, as indicated by `-> str`.

```python
def get_user() -> dict:
    return {"id": 101, "name": "Akshay", "email": "akshay@example.com"}
```

```python
from typing import TypedDict

class UserResponse(TypedDict):
    id: int
    name: str
    email: str

def get_user() -> UserResponse:
    return {"id": 101, "name": "Akshay", "email": "akshay@example.com"}
```

```python
for key, value in get_user().items():
    print(f"Type of {key}: {type(value)}, Entity: {value}")
```
