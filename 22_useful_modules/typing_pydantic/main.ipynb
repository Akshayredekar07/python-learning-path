## **typing — Support for type hints**

### **Added in version 3.5**

Source code: `Lib/typing.py`

**Note**: The Python runtime does not enforce function and variable type annotations.  
They can be used by third-party tools such as type checkers, IDEs, linters, etc.  
This module provides runtime support for type hints.


### **Example: Function with Type Hints**

```python
def surface_area_of_cube(edge_length: float) -> str:
    return f"The surface area of the cube is {6 * edge_length ** 2:.2f}."
```

```python
result = surface_area_of_cube(4.4)
result
```

```python
type(result)
```


* The function `surface_area_of_cube` takes an argument expected to be a **float**.
* It is expected to return a **str**, as indicated by `-> str`.

```python
def get_user() -> dict:
    return {"id": 101, "name": "Akshay", "email": "akshay@example.com"}
```

```python
from typing import TypedDict

class UserResponse(TypedDict):
    id: int
    name: str
    email: str

def get_user() -> UserResponse:
    return {"id": 101, "name": "Akshay", "email": "akshay@example.com"}
```

```python
for key, value in get_user().items():
    print(f"Type of {key}: {type(value)}, Entity: {value}")
```

```python
class PredictionResult(TypedDict):
    prediction: str   # "REAL" or "FAKE"
    confidence: float
    model_version: str

def run_inference(video: str) -> PredictionResult:
    return {"prediction": "FAKE", "confidence": 0.98, "model_version": "v2.3"}
```

### **Type Aliases**

A type alias is defined using the `type` statement, which creates an instance of `TypeAliasType`.
In this example, `Vector` and `list[float]` will be treated equivalently by static type checkers:

```python
type Vector = list[float]

def scale(scalar: float, vector: Vector) -> Vector:
    return [scalar * num for num in vector]

# passes type checking
new_vector = scale(2.0, [1.0, -4.2, 5.4])
```

```python
new_vector
```

```python
# Use Case: Simplifying complex type signatures.

from collections.abc import Sequence

type ConnectionOptions = dict[str, str]
type Address = tuple[str, int]
type Server = tuple[Address, ConnectionOptions]

def broadcast_message(message: str, servers: Sequence[Server]) -> None:
    for server in servers:
        address, options = server
        host, port = address
        print(f"Sending '{message}' to {host}:{port} with options {options}")

# Example usage
servers = [
    (("localhost", 8080), {"timeout": "30s", "retry": "3"}),
    (("remote.example.com", 443), {"ssl": "true", "auth": "token"})
]
broadcast_message("Hello, world!", servers)
```

```python
def broadcast_message(
    message: str,
    servers: Sequence[Server]
) -> None:
    for server in servers:
        address, options = server
        host, port = address
        print(f"Sending '{message}' to {host}:{port} with options {options}")
```

### **NewType**

Use the `NewType` helper to create distinct types:

```python
from typing import NewType

UserId = NewType('UserId', int)
some_id = UserId(524313)
```

```python
some_id
```

```python
def get_user_name(user_id: UserId) -> str:
    return f"User {user_id}"

# passes type checking
user_a = get_user_name(UserId(42351))

# passes type checking; use UserId to wrap the int
user_b = get_user_name(UserId(-1))
```

```python
user_a, user_b
```
